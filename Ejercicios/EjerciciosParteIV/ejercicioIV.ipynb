## 7. Más ejemplos de coincidencia de argumentos
Primero, defina las siguientes seis funciones (ya sea de forma interactiva o en un archivo de módulo que se pueda importar).
Ahora, pruebe las siguientes llamadas de forma interactiva e intente explicar cada resultado. ¿Cree que mezclar modos de coincidencia es una buena idea en general? ¿Puede pensar en casos en los que sería útil?


```python
def f1(a, b): print(a, b) # Argumentos normales
def f2(a, *b): print(a, b) # Recolectores posicionales
def f3(a, **b): print(a, b) # Recolectores de palabras clave
def f4(a, *b, **c): print(a, b, c) # Modos mixtos
def f5(a, b=2, c=3): print(a, b, c) # Valores predeterminados
def f6(a, b=2, *c): print(a, b, c) # Valores predeterminados y recolectores posicionales

f1(1, 2)
f1(b=2, a=1)
f2(1, 2, 3)
f3(1, x=2, y=3)
f4(1, 2, 3, **dict(x=2, y=3))
f5(1)
f5(1, 4)
f5(1, c=4)
f6(1)
f6(1, *[3, 4])
```
