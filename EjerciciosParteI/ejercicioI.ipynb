# Configuración, Entorno y Ejecución Inicial

## 1. Interacción
Usando una línea de comandos del sistema, IDLE o cualquier otro método que funcione en su dispositivo, inicie la línea de comandos interactiva de Python (el indicador >>>, también conocido como REPL) y escriba la expresión 'Hello World!' (incluyendo las comillas). La cadena debería devolverse reflejada. El propósito de este ejercicio es configurar su entorno para ejecutar Python. En escenarios raros, es posible que deba escribir la ruta completa al ejecutable de Python o agregar su ruta a la variable de entorno PATH. Consulte el Apéndice A para obtener consejos sobre la configuración de variables de entorno si es necesario.


## 2. Programas
Con el editor de texto de su elección, escriba un archivo de módulo simple que contenga la declaración única print('Hello module world!') y guárdelo como module1.py. Ahora, ejecute este archivo utilizando cualquier opción de inicio que desee: ejecutándolo en IDLE, haciendo clic o tocando el icono de su archivo, iniciándolo por línea de comandos en una consola (por ejemplo, python3 module1.py), ejecutándolo en código con herramientas como exec e imports/reloads, o mediante opciones de interfaz de usuario en aplicaciones, otros IDEs y cuadernos web. De hecho, experimente ejecutando su archivo con tantas técnicas de inicio analizadas en este capítulo como pueda. ¿Qué técnica le parece más fácil?

## 3. Módulos
Inicie la línea de comandos interactiva de Python (indicador >>>) e importe el módulo que escribió en el ejercicio 2. Intente mover el archivo a un directorio diferente e impórtelo nuevamente desde su directorio original (es decir, ejecute Python en el directorio original cuando realice la importación). ¿Qué pasa? Pista: todavía hay un archivo de código de bytes .pyc para module1 en un subdirectorio __pycache__ allí, pero tiene un nombre extraño. En general, las importaciones buscan módulos en el directorio actual, más cada directorio listado en la variable de entorno PYTHONPATH.

## 4. Scripts
Si su plataforma lo admite, copie su archivo de módulo module1.py a otro llamado script1.py; luego agregue la línea #! en la parte superior de script1.py, otorgue a este archivo privilegios de ejecutable y ejecútelo directamente como un ejecutable (por ejemplo, sin el "python3"). ¿Qué debe contener la primera línea? Las líneas #! tradicionalmente tienen significado solo en plataformas tipo Unix (por ejemplo, macOS, Linux y Android), pero también se aplican a Windows hoy en día, gracias al lanzador py de Windows. Si está trabajando en Windows, intente también ejecutar su archivo enumerando solo su nombre en una ventana de símbolo del sistema sin la palabra py antes, a través del menú Inicio, el cuadro de diálogo Ejecutar Windows+R u otros esquemas. En macOS, intente arrastrar y soltar en la aplicación Python Launcher en Aplicaciones (o en cualquier otro lugar).

## 5. Errores y depuración
Experimente escribiendo expresiones matemáticas y asignaciones en la línea de comandos interactiva de Python (es decir, REPL). En el camino, escriba las expresiones 2 ** 500 y 1 / 0, y haga referencia a un nombre de variable indefinido como lo hicimos al principio de este capítulo. ¿Qué pasa?
Es posible que aún no lo sepa, pero cuando comete un error, está realizando un procesamiento de excepciones, un tema que exploraremos en profundidad en la Parte VII. Como aprenderá allí, técnicamente está activando lo que se conoce como el manejador de excepciones predeterminado: lógica que imprime un mensaje de error estándar. Si no captura un error, el manejador predeterminado lo hace e imprime el mensaje de error estándar en respuesta.
Las excepciones también están vinculadas a la noción de depuración en Python. Cuando recién está comenzando, los mensajes de error predeterminados de Python en las excepciones probablemente le brindarán todo el soporte de manejo de errores que necesita: indican la causa del error, además de mostrar las líneas en su código que estaban activas cuando ocurrió el error. Para obtener más información sobre la depuración, consulte la próxima barra lateral "Depuración de código Python".
