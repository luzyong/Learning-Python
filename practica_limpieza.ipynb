## 🔍 Paso 2: Análisis de Claves Dinámicas y Anidación
Para limpiar un diccionario, primero debemos entender los niveles en los que están guardados sus elementos. En este paso, extraeremos las claves del diccionario principal, sus valores y descenderemos al nivel donde se encuentra el contenido de la receta.

A partir del diccionario que obtuviste en el paso anterior, obtén las llaves externas del diccionario y sus valores y todas las llaves internas del diccionario. Guardalas en una lista e imprimelas para explorarlas.

```python
keys_0 = list(dictionary.keys())
values_0 = list(dictionary.values())
keys_1 = list(values_0[0][0].keys())

print("Claves de primer nivel (keys_0):", keys_0)
print("\nClaves internas de la receta (keys_1):\n", keys_1)
```

## 🧼 Paso 3: Filtrado y Limpieza Avanzada de Datos
La respuesta de la API incluye campos fijos para hasta 20 ingredientes (como `strIngredient1`, `strIngredient2`, etc.), pero la mayoría de las recetas usan menos. Esto genera claves que contienen cadenas vacías `""`, espacios en blanco `" "` o valores nulos (`None`). 

Utilizaremos la función integrada `filter()` junto con expresiones `lambda` para aislar únicamente las claves que **sí contienen datos válidos**, separándolos por categorías (Ingredientes, Medidas y Otros).

**Una función lambda es una función anónima que ejecuta código de una sola línea. Su sintaxis es la siguiente: lambda x: hacer algo con x**

Asegúrate de utilizar una sola lista para filtrar las categorías mencionadas anteriormente asegurandote de que solamente estén las llaves que tienen alguna información.

```python
clean = []

# 1. Filtrar claves que contengan 'Ingredient' y que su valor no esté vacío o sea un espacio
clean = list(filter(lambda x: "Ingredient" in x and (dictionary[keys_0[0]][0][x] and dictionary[keys_0[0]][0][x] != " "), keys_1))

# 2. Filtrar y añadir claves que contengan 'Measure' y que no estén vacías
clean += list(filter(lambda x: "Measure" in x and (dictionary[keys_0[0]][0][x] and dictionary[keys_0[0]][0][x] != " "), keys_1))

# 3. Filtrar y añadir el resto de las claves operativas (Id, Nombre de receta, Instrucciones, etc.) que tengan contenido válido
clean += list(filter(lambda x: "Measure" not in x and "Ingredient" not in x and (dictionary[keys_0[0]][0][x] and dictionary[keys_0[0]][0][x] != " "), keys_1))

print("Lista de claves limpias (clean):\n", clean)
```
