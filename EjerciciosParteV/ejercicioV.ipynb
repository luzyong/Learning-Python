## 5. Importaciones de paquetes
Importe su archivo mymod.py desde un paquete. Cree un subdirectorio llamado mypkg anidado en un directorio en su ruta de búsqueda de importación de módulos, copie o mueva el archivo de módulo mymod.py que creó en el ejercicio 1 o 3 al nuevo directorio e intente importarlo con una importación de paquete de la forma import mypkg.mymod y llame a sus funciones. Intente buscar sus funciones de contador con un from también.
Esto funciona en todas las plataformas de Python (esa es parte de la razón por la que Python usa "." como separador de ruta). El directorio de paquetes que cree puede ser simplemente un subdirectorio de aquel en el que está trabajando; si lo es, se encontrará a través del componente del directorio de inicio de la ruta de búsqueda y no tendrá que configurar su ruta.
Tampoco necesita un archivo __init__.py en el directorio del paquete al que se movió su módulo para que esto funcione, pero cree uno con algunas impresiones básicas y vea si se ejecutan en cada importación o recarga de la carpeta del paquete. Finalmente, copie también mymod.py a la carpeta del paquete __main__.py e invoquelo ejecutando la carpeta misma; ¿tiene sentido hacer eso aquí? ¿Aún puede ejecutar el módulo anidado mymod.py propiamente dicho?

## 7. Importaciones circulares
En la sección sobre errores de importación recursiva (también conocidos como circulares), la importación de recur1 generó un error. Pero si reinicia Python e importa recur2 de forma interactiva, el error no ocurre; pruebe esto y compruébelo usted mismo. ¿Por qué cree que funciona importar recur2 pero no recur1? (Pista: Python registra módulos nuevos antes de ejecutar su código, y las importaciones posteriores buscan el módulo primero, ya sea que el módulo esté "completo" o no).
Ahora, ejecute recur1 como un archivo de script de nivel superior: python3 recur1.py. ¿Obtiene el mismo error que ocurre cuando recur1 se importa de forma interactiva? ¿Por qué? (Pista: cuando los módulos se ejecutan como programas, no se importan, por lo que este caso tiene el mismo efecto que importar recur2 de forma interactiva; recur2 es el primer módulo importado). ¿Qué sucede cuando ejecuta recur2 como un script? Las importaciones circulares son poco comunes en la práctica. Por otro lado, si puede entender por qué son un problema potencial, sabrá mucho sobre la semántica de importación de Python.
