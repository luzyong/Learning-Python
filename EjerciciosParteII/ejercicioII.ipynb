## Números.
<p style="text-align: justify; font-style: italic;">
Los números son un tipo de dato inmutable que soporta operaciones aritméticas, de comparación, de bitwise. Los numeros hacen una conversion implicita al hacer operaciones entre diferentes subtipos, por lo que se convierten al subtipo de numero con más peso complejo>float>entero.
</p>

```python
print(2**6)
print(2/5, 2/5.0)
```

## Strings y Formatting.
<p style="text-align: justify; font-style: italic;">
Las strings son un tipo de dato inmutable y de la clase secuencia. Esto significa que puede ser indexada y soporta slicing. También soporta operaciones como la concatenación, que es la suma de dos o más strings y la multiplicación que puede ser entendida como la suma de una string n veces. Podemos hacer formateo para imprimir valores de variables u objetos dentro de una cadena de texto.
</p>


```python
print("spam"+"eggs")

s="ham"
print("eggs " + s)
print(s*5)
print(s[:0])

print("green %s ans %s" % ("eggs",s))
print('green {0} and {1}'.format('egss', s))
```

## Secuencias. Listas y tuplas.
<p style="text-align: justify; font-style: italic;">
Las secuencias soportan operaciones similares como la indexacion y el slicing, la suma y la multiplicación. 
Las listas son tipos de secuencías mutables, por lo que pueden ser cambiadas in-place o actualizadas. Esto afecta el valor del objeto y por consiguiente a todas las variables que hacen referencia a él.
Las tuplas son tipos de secuencias inmutables, por lo que no pueden ser cambiadas in-place o actualizadas. Estas no soportan las operaciones que hacen este tipo de cambios.
Indexamos una secuencia al proporcionarle a un par de corchetes una posición de la lista que va de 0 a tamaño-1. Al indexar una lista obtenemos el elemento individual que esta en la posición determinada.
Hacemos un slice cuando queremos obtener un subconjunto de una lista y la determinamos al proporcional el rango del que vamos a extraer ese conjunto. El rango inferior es inclusivo y el superior es exclusivo.
Una asignacion de indice cambia el elemento en la posicion indicada.
Una asignación de slice elimina e inserta una porcion de listo dentro de otra.
</p>

```python
print(('x',)[0])
print(('x','y')[1])

L = [1,2,3] + [4,5,6]
print(L)
print(L[:], L[:0], L[-2], L[-2:])
print(([1,2,3] + [4,5,6])[2:4])
print([L[2], L[3]])
L.reverse()
print(L)
L.sort()
print(L)
print(L.index(4))
```


## Diccionarios.
<p style="text-align: justify; font-style: italic;">
Un diccionario es un tipo de dato de la clasificación mapping y es mutable. Su funcionamiento se basa en clave-valor, que permite extraer elementos a partir de una llave en vez de una posición, lo que lo hace más eficiente.</p>

```python
print({'a':1, 'b':2}['b'])
D = {'x':1, 'y':2, 'z':3}
D['w'] = 0
print(D['x'] + D['w'])
D[(1,2,3)] = 4
print(list(D.keys()), list(D.values()), (1,2,3) in D)
print([[]], ["",[],(),{},None])
```

# Indexing y Slicing


Prueba este bloque para observar errores y comportamientos especiales:

```python
L = [0,1,2,3]

# Index out of bounds
try:
    print(L[4])
except Exception as e:
    print("Error en L[4]:", e)

# Slice out of bounds
print("Slice fuera de rango:", L[-1000:100])

# Slice invertido
print("Slice invertido:", L[3:1])
L[3:1] = ['?']
print("Después de asignar a slice invertido:", L)
```

# Indexing, Slicing y del

```python
L = ['a','b','c','d']
L[2] = []
print("Después de L[2] = []:", L)

L = ['a','b','c','d']
L[2:3] = []
print("Después de L[2:3] = []:", L)

L = ['a','b','c','d']
del L[0]
print("Después de del L[0]:", L)

L = ['a','b','c','d']
del L[1:]
print("Después de del L[1:]:", L)

L = ['a','b','c','d']
L[1:2] = [1] # Para evitar errores, se mete en lista
print("Después de L[1:2] = 1:", L)
```
